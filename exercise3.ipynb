```python
#list operations
L1 =[10,20,30]
L2 =[40,50,60]

#1.concatentation
print(L1 + L2)

#2.repetition
print(L2 * 3)

#3.indexing
print(L1[2])

#4.slicing
print(L1[0:2])

#5.membership
print(20 in L1)

#iteration
print
for i in L1:
  print(i)
```

```python
#list functions
a=[10,20,30,40,50]

print(sum(a))
print(max(a))
print(min(a))
print(len(a))
a.reverse()
print(a)
print(sorted(a))
```

```python
#list methods
L=[10,20,30,40,50]
#Append
L.append(50)
print("append:",L)
#Extend
L.extend([60,70])
print("extend:",L)
#Insert
L.insert(1,15)
print("insert:",L)
#pop
L.pop()
print("pop:",L)
#remove
L.remove(30)
print("remove:",L)
#index
L.index(20)
print("index:",L)
#count
L.count(30)
print("count:",L)
#reverse
L.reverse()
print("reverse:",L)
#sort
L.sort()
print("sort:",L)
```

```python
#program that prints sum of lists elements without using sum()
L=[10,20,30,40,50]
total= 0
for i in L:
  total=total +i
print("sum =",total)
```

```python
#print the list elements at odd index
L=[10,20,30,40,50,60]
print("elements at odd index:")
for i in range(1,len(L),2):
  print(L[i], end=" ")
```

```python
# program on list compression
L =[1,2,3,4,5]
square =[i*i for i in L]
print("orginal list:",L)
print("square list:",square)
```

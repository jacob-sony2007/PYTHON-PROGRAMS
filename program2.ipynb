```python
# program to find the date of given number
day=int(input())
match day:
  case 1:
    print("sunday")
  case 2:
      print("monday")
  case 3:
      print("tuesday")
  case 4:
      print("wednesday")
  case 5:
      print("thursday")
  case 6:
      print("friday")
  case 7:
      print("saturday")
  case _:
      print("invalid")
```

```python
# program that finds sum of digits of a given number
n=int(input("enter a number"))
sum=0
while(n>0):
  r=n%10
  sum=sum+r
  n=n//10
  print("sum of the given number is",sum)
```

```python
# program that checks given number is prime or not
n=int(input("enter a number"))
for i in range(2,n):
  if(n%i==0):
    print("not a prime")
    break
  else:
    print("prime number")
```

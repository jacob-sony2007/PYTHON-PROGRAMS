```python
#write a program to check if a given key exists in a dictionary or not
my_dict = {"name":"jacob","age":19,"course":"btech"}
key = input("enter a key:")
if key in my_dict:
  print("key exist in the dictionary")
else:
  print("key does not exist in the dictionary")
```

```python
#write a program to add a new key value pair to an existing dictionary
d = {"name":"jacob","age":19,}
d["city"] ="chinnaamiram"
print(d)
```

```python
# write a program to sum all the items in a given dictionary
d = {'a':10, 'b': 20, 'c': 30}
sum = 0
for value in d.values():
  sum = sum + value
  print("sum =", sum)
```

```python
#write a program that prints frequency of each  element in a list in following
a =[1 ,2 , 3, 4, 5, 6, 1, 6, 7]
dic ={}
for i in a:
  if i in dic:
    dic[i] = dic[i] + 1
  else:
    dic[i] = 1
print(dic)
```
